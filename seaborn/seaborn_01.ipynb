```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
tips.head()
```

```python
sns.relplot(
    data = tips,
    x= "total_bill",
    y= "tip",
    hue ="smoker",
    size = "size",
    style="smoker"
)
```

```python
x= [1,2,3,4,5,6,7,8,9]
y1 = [i**2 for i in x]

sns.relplot(
    x= x,
    y=y1,
    kind = "line",
    marker = "o",
    color = "red"
)
```

```python
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
    
)
```

```python
flights = sns.load_dataset("flights")
flights.head()
sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers",
    marker = "o"
)
```

```python
sns.barplot(
    data = tips,
    x= "day",
    y= "tip",
    hue="sex"
)
```

```python
sns.boxplot(
    data = tips,
    x = "day",
    y= "tip"
)
```

```python
penguins = sns.load_dataset("penguins")

sns.histplot(
    data = penguins,
    x= "body_mass_g",
    bins = 20
)
```

```python
flights.head()
```

```python
flights_pivot = flights.pivot(index = "month", columns = "year", values = "passengers")

sns.heatmap(
    flights_pivot,
    cmap = "coolwarm",
    annot= True,
    fmt = "d"
)
plt.tight_layout()
plt.title("passengers heatmap")
```

```python
fig, ax = plt.subplots()

sns.lineplot(
    data = tips,
    x= "day",
    y ="total_bill",
    hue = "sex",
    ax = ax,
    marker="o",
    errorbar = None
)
ax.set_title("avg bill value for each day")
```
