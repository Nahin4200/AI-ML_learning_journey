```python
import numpy as np
import time
```

```python
# Python list

size = 1_000_000

py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print (f"python list time = {end - start}seconds")

#numpy array
np_array = np.array(py_list)
start = time.time()
sq_array = np_array **2
end = time.time()
print (f"numpy array time = {end - start}seconds")
```

```python
#memory 
import sys

print (f"python list memory size = {sys.getsizeof (py_list) * len (py_list)} bytes")

print (f"numpy array memory size = {np_array.nbytes} bytes")
```
