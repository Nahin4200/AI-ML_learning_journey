```python
people = ["person A", "person B", "person C", "person D", "person E",
         "person F", "person G", "person H", "person I", "person J"
         ]
age = [22, 25, 30,35,40,45,50,55,60,65]
bp = [110,115,120,122,125,130,135,123,145,150]
colors = ["green" if x <130 else "red" for x in bp]

# plt.scatter(age, bp, color = colors, s=bp, alpha= 0.7)
plt.scatter(age, bp, s=bp, cmap="OrRd", c=bp)
plt.title("Age vs blood_pressure")
plt.xlabel("Age")
plt.ylabel("Blood_pressure")
plt.grid()
plt.colorbar(label = "BP")
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i],bp[i]), xytext = (age[i]+1,bp[i]+1))

plt.xlim(min(age)-3, max(age)+10)
plt.ylim(min(bp)-2, max(bp)+4)
```

```python
#sub plots
x = [1,2,3,2,4,5,6,7,8,9,10]

y1 = [np.sqrt(i) for i in x] #square root
y2 = [i*2 for i in x]
y3 =[i**2 for i in x]
y4 = [i**3 for i in x]

plt.subplot(2,2, 1)
plt.plot(x, y1)
plt.title("square root")
plt.grid()

plt.subplot(2,2, 2)
plt.plot(x, y2)
plt.title("doubles")
plt.grid()

plt.subplot(2,2, 3)
plt.plot(x, y3)
plt.title("squares")
plt.grid()

plt.subplot(2,2, 4)
plt.plot(x, y4)
plt.title("cubes")
plt.grid()
plt.tight_layout()
```

```python
fig, ax = plt.subplots(2,2)

ax[0][0].plot(x,y1)
ax[0][0].set_title("square root")
ax[0][0].set_xlabel("x_axis_value")
ax[0][0].set_ylabel("y_axis_valu")


ax[0][1].plot(x,y2, color = 'black')
ax[0][1].set_title("square root")
ax[0][1].set_xlabel("x_axis_value")
ax[0][1].set_ylabel("y_axis_valu")

ax[1][0].plot(x,y3, color = "green")
ax[1][0].set_title("square root")
ax[1][0].set_xlabel("x_axis_value")
ax[1][0].set_ylabel("y_axis_valu")

ax[1][1].plot(x,y4, color = "red")
ax[1][1].set_title("square root")
ax[1][1].set_xlabel("x_axis_value")
ax[1][1].set_ylabel("y_axis_valu")
fig.tight_layout()
```
