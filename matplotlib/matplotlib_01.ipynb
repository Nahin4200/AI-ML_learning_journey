```python
non_oscar_movies = ["abc", "de", "ef", "asd", "fgh"]
non_oscar_revenue = [904,2006,1100,2204,3829]

plt.plot(year, oscar_revenue, "o--r", label ="oscar_movies" )
plt.plot(year, non_oscar_revenue, color="blue", marker="o", linestyle="--", label = "non_oscar_movies")
plt.title("oscar movies vs non-oscar movies revenue ")
plt.xlabel("year")
plt.ylabel("oscar_revenue (in $)")
plt.legend()
plt.grid()
```

```python
import numpy as np
oscar_movies = [
    "Nomadland",
    "Parasite",
    "Green Book",
    "The Shape of Water",
    "Moonlight"  
]

years = [2016,2015,2014,2013,2012]
oscar_revenue = [204,1006,1800,804,1329]

non_oscar_movies = ["kgf", "RRR", "Crash", "The Departed", "got"]
non_oscar_revenue = [904,2006,1100,2204,3829]

x= np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label = "oscar_movies")
plt.bar(x+width/2, non_oscar_revenue, width, label = "non_oscar_movies")

plt.title("oscar movies vs non oscar movies revenue")
plt.xlabel("year")
plt.ylabel("Revenue ($M)")
plt.legend()
plt.xticks(x, years)
plt.savefig("movies.png")
```

```python
people = ["person A", "person B", "person C", "person D", "person E",
         "person F", "person G", "person H", "person I", "person J"
         ]
age = [22, 25, 30,35,40,45,50,55,60,65]
bp = [110,115,120,122,125,130,135,123,145,150]
colors = ["green" if x <130 else "red" for x in bp]

# plt.scatter(age, bp, color = colors, s=bp, alpha= 0.7)
plt.scatter(age, bp, s=bp, cmap="OrRd", c=bp)
plt.title("Age vs blood_pressure")
plt.xlabel("Age")
plt.ylabel("Blood_pressure")
plt.grid()
plt.colorbar(label = "BP")
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i],bp[i]), xytext = (age[i]+1,bp[i]+1))

plt.xlim(min(age)-3, max(age)+10)
plt.ylim(min(bp)-2, max(bp)+4)
```
