```python
# Handle string

df["gender"].str.lower()
df["gender"].str.upper()
df["gender"].str.capitalize()

df["name"].str.split(" ")

df["country"].str.contains("US")
df["country"].str.contains("india", case = False)
```

```python
#Transformation of the data

df2 = df.copy()
df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 50000 else "10%") 


gender_map = {"Male": "M", "Female": "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(gender_map)


df2 = df2.assign(new_income = df2["income"] * 1.1)
df2


df2["country"] = df2["country"].replace("USA", "US")


df2.columns = ["ID", "NAME", "AGE", "COUNTRY", "GENDER", "INCOME", "TAX", "NEW_INCOME"]
df2 = df2.rename(columns = {"INCOME":"SELARY"})
df2 = df2.rename(columns = {"SELARY":"INCOME"})


df2.rename(index = {1: "first"})

df2.sort_values("INCOME")
# df2.sort_values("INCOME", ascending = False)
sorted_data = df2.sort_values(["INCOME","AGE"])
# sorted_data.sort_index()
sorted_data.reset_index()
sorted_data.reset_index(drop = True)

sorted_data["Ranking"] = sorted_data["INCOME"].rank(ascending=False, method= "dense")
sorted_data
```

```python
#Group by & aggregation

df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].max()

df.groupby("gender")["income"].agg(["mean", "min", "max"])
df.groupby("gender")["income"].agg(avg_salary="mean", min_salary="min", max_salary="max")

df.groupby("country").agg({
    "income" : "mean",
    "age" : "min"
})

df.groupby("country").agg({
    "income" : "max",
    "age" : "min"
})


df.groupby("country").agg(
   max_salary = ("income", "max"),
    min_age = ("age", "min")
)
```

```python
#melt & pivot

df = pd.DataFrame({
    "country" : ["US", "US", "BAN", "BAN"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" : [100, 120, 90, 110],
    "profit" : [20, 25, 18, 22]
    
})

melt_df = df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit"],
    var_name = "matrics",
    value_name = "value"
)
melt_df.pivot(
    index = ["country", "year"],
    columns = "matrics",
    values = "value"
)
```
