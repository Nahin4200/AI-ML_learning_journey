```python
import json
```

```python
def load_data (filename):
    with open(filename, "r") as f:
        data = json.load(f)

        return data
        
```

```python
#load the data
data = load_data ("store_data.json")
print (data)
print (type(data))
```

```python
# Clean & structure the data

def clean_data (data):
    #cleaning rating
    txt_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []
    unique_users = set ()
    
    
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if( raw_rating in txt_to_num):
            raw_rating = txt_to_num[raw_rating]
        user ["rating"] = raw_rating
        
        #handle missing vals
        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None

       # deduplication
        if (user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"]) 

        cleaned_data.append(user)
           
    return cleaned_data       
           
    
        
```

```python
clean_data (data)
```

```python
# get meaningful insights from data
data =  clean_data(data)


def get_insights(data):
    total_rating = 0

    #avg rating
    for user in data:
        total_rating += float(user["rating"])
    print (f" Avg rating = {total_rating / len(data)}")

    # % of users with poor rating
    poor_rating = 0
    for user in data:
        if (float (user["rating"]) < 3):
            poor_rating +=1
    print (f"% of users with poor rating = {(poor_rating / len(data)) * 100}%")
            
    

```

```python
get_insights(data)
```

```python
#Build recommendation feature
def get_recommendation(data):
    recommendation = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user ["name"]
        
        if(float (user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "samsung"
            
        recommendation.append(curr_recomm)

    return recommendation
            
    
```

```python
get_recommendation(data)
```
