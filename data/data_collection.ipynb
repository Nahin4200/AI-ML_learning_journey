```python
url = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(url)
```

```python
res
```

```python
if res.status_code == 200:
    print(res.headers)

with open("scrape_data/data1.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
pip install lxml
```

```python
from bs4 import BeautifulSoup

with open("scrape_data/data1.html", "r") as f:
    html_content = f.read()
    
soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h1")
#soup.h3
```

```python
#soup.find_all("h3")
all_h3 = soup.find_all("h3")

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)
    # print(h3.find_parent("div")["class"])
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    capital = h3.find_next("div").select("span.country-capital")[0].get_text(strip = True)
    all_countries.append([name, capital, population])
```

```python
import pandas as pd

df = pd.DataFrame(all_countries, columns=["name", "capital", "population"])
df
```

```python
df.to_csv("cleaned_data/data1.csv", index=False)
```
